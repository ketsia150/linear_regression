## Importing dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import sklearn 
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (accuracy_score, confusion_matrix, classification_report, ConfusionMatrixDisplay)
```

## Data Processing

```python
#Check the target balance
print(df['result'].value_counts())

#Drop the redundant column
df2 = df.drop(columns=['sum'])

#Encode the target (yes/no -> 1/10)
le = LabelEncoder()
y = le.fit_transform(df2['result'])
x = df2.drop(columns = ['result'])

# Split (stratify keps the yes/no ratio the same)
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size= 0.2, random_state= 42, stratify = y)

# Normalize (the assignement asks for it, though trees don't need it)
scaler = MinMaxScaler()
x_train_s = scaler.fit_transform(x_train) 
x_test_s = scaler.transform(x_test)
```

## Modeling

```python
dt = DecisionTreeClassifier(random_state= 42)
params = {'max_depth': [3,5,7, 10, None],
          'min_samples_split': [2,5,10],
          'criterion': ['gini', 'entropy']}

grid = GridSearchCV(dt, params, cv = 5, scoring='accuracy')
grid.fit(x_train_s, y_train)

model = grid.best_estimator_
print(grid.best_params_)
```

## Evaluation

```python
y_pred = model.predict(x_test_s)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred, target_names = le.classes_))

ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred),
                       display_labels= le.classes_).plot()
plt.show()

#Which features matter most
print(pd.Series(model.feature_importances_, index= x.columns).sort_values(ascending = False))

# Visualize the tree
plt.figure(figsize = (16,8))
plot_tree(model, feature_names = x.columns, 
          class_names = le.classes_, filled=True, max_depth=3)
plt.show()
```

```python
print("Train accuracy:", model.score(x_train_s, y_train))
print("Test accuracy:", model.score(x_test_s, y_test))
```

```python
## Target variable: `result` (yes/no), encoded as 1/0.
## Data cleaning: There were no missing values or duplicates. I dropped `sum` because it is just the other columns added together and has an outlier.
## processing: I split the data 80/20 with stratify, and normalized with MinMaxScaler (decision trees don't need scaling, so it doesn't change the results).
## Model: Decision Tree with GridSearchCV. The best parameters were criterion = entropy, max_depth = 7, min_samples_split = 2.
## Performance: Test accuracy was 94%, with precision, recall and f1 around 0.94 for both classes. The confusion matrix shows 100 correct "No", 89 correct "yes", 4 false "yes" and 8 false "No".
## Feature importance: `last_payment` is the strongest predictor (0.71). `house_number` (0.19) is odd because it is basically an ID, so it may not be a meaningful feature.
```
